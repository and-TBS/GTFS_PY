```python
import os
import pandas as pd
import geopandas as gpd
from datetime import datetime, timedelta
import folium
import ipywidgets as widgets
from IPython.display import display, clear_output

# ==========================================
# CONFIGURAZIONE FILE
# ==========================================
GTFS_DIR = "."
GEOJSON_FILE = "Italia.geojson"  # Nome aggiornato del file GeoJSON
OUTPUT_EXCEL = "riepilogo_gtfs_definitivo.xlsx"

# ==========================================
# 1. CARICAMENTO ED ELABORAZIONE DATI GTFS
# ==========================================
print("🚀 [1/4] Caricamento ed elaborazione dati GTFS...")

routes = pd.read_csv(os.path.join(GTFS_DIR, 'routes.txt'))
trips = pd.read_csv(os.path.join(GTFS_DIR, 'trips.txt'))
stop_times = pd.read_csv(os.path.join(GTFS_DIR, 'stop_times.txt'))
stops = pd.read_csv(os.path.join(GTFS_DIR, 'stops.txt'))

calendar_path = os.path.join(GTFS_DIR, 'calendar.txt')
calendar_dates_path = os.path.join(GTFS_DIR, 'calendar_dates.txt')
has_calendar = os.path.exists(calendar_path)

# --- Pulizia fermate per evitare duplicati ID ---
stops_unique = stops.drop_duplicates(subset=['stop_id'])

# --- Calcolo Capolinea e Orari unici per trip_id ---
stop_times['stop_sequence'] = pd.to_numeric(stop_times['stop_sequence'])
stop_times = stop_times.sort_values(['trip_id', 'stop_sequence'])

first_stops = stop_times.groupby('trip_id').first().reset_index()[['trip_id', 'stop_id', 'departure_time']]
first_stops.rename(columns={'stop_id': 'start_stop_id'}, inplace=True)

last_stops = stop_times.groupby('trip_id').last().reset_index()[['trip_id', 'stop_id', 'arrival_time']]
last_stops.rename(columns={'stop_id': 'end_stop_id'}, inplace=True)

trip_endpoints = pd.merge(first_stops, last_stops, on='trip_id')

# Unione nomi fermate con dataset pulito
trip_endpoints = pd.merge(trip_endpoints, stops_unique[['stop_id', 'stop_name']], left_on='start_stop_id', right_on='stop_id', how='left')
trip_endpoints.rename(columns={'stop_name': 'fermata_partenza'}, inplace=True)
trip_endpoints.drop(columns=['stop_id', 'start_stop_id'], inplace=True)

trip_endpoints = pd.merge(trip_endpoints, stops_unique[['stop_id', 'stop_name']], left_on='end_stop_id', right_on='stop_id', how='left')
trip_endpoints.rename(columns={'stop_name': 'fermata_arrivo'}, inplace=True)
trip_endpoints.drop(columns=['stop_id', 'end_stop_id'], inplace=True)

trip_full = pd.merge(trip_endpoints, trips, on='trip_id', how='left')

if 'route_short_name' in routes.columns:
    routes['linea'] = routes['route_short_name'].fillna(routes['route_long_name'])
else:
    routes['linea'] = routes['route_long_name']

trip_full = pd.merge(trip_full, routes[['route_id', 'linea']], on='route_id', how='left')

# --- CALCOLO PRECISO GIORNI DI VALIDITÀ (Calendar + Calendar Exceptions) ---
service_active_days = {}
service_dates_range = {}

if has_calendar:
    calendar = pd.read_csv(calendar_path).drop_duplicates(subset=['service_id'])
    days_cols = ['monday', 'tuesday', 'wednesday', 'thursday', 'friday', 'saturday', 'sunday']
    
    for _, row in calendar.iterrows():
        s_id = row['service_id']
        try:
            start = datetime.strptime(str(int(row['start_date'])), '%Y%m%d')
            end = datetime.strptime(str(int(row['end_date'])), '%Y%m%d')
            service_dates_range[s_id] = (row['start_date'], row['end_date'])
            
            active_dates = set()
            curr = start
            while curr <= end:
                if row.get(days_cols[curr.weekday()]) == 1:
                    active_dates.add(curr)
                curr += timedelta(days=1)
            service_active_days[s_id] = active_dates
        except Exception:
            service_active_days[s_id] = set()

if os.path.exists(calendar_dates_path):
    calendar_dates = pd.read_csv(calendar_dates_path)
    for _, row in calendar_dates.iterrows():
        s_id = row['service_id']
        try:
            d_str = str(int(row['date']))
            dt = datetime.strptime(d_str, '%Y%m%d')
            exc_type = row['exception_type'] # 1 = aggiunto, 2 = rimosso
            
            if s_id not in service_active_days:
                service_active_days[s_id] = set()
            
            if exc_type == 1:
                service_active_days[s_id].add(dt)
            elif exc_type == 2:
                service_active_days[s_id].discard(dt)
        except Exception:
            pass

def get_service_info(service_id, field):
    if service_id not in service_active_days:
        return 0 if field == 'giorni' else ''
    if field == 'giorni':
        return len(service_active_days[service_id])
    elif field == 'start':
        dates = service_active_days[service_id]
        if dates:
            return min(dates).strftime('%Y%m%d')
        return service_dates_range.get(service_id, ('', ''))[0]
    elif field == 'end':
        dates = service_active_days[service_id]
        if dates:
            return max(dates).strftime('%Y%m%d')
        return service_dates_range.get(service_id, ('', ''))[1]

trip_full['giorni_validita'] = trip_full['service_id'].apply(lambda x: get_service_info(x, 'giorni'))
trip_full['start_date'] = trip_full['service_id'].apply(lambda x: get_service_info(x, 'start'))
trip_full['end_date'] = trip_full['service_id'].apply(lambda x: get_service_info(x, 'end'))

# Eliminazione rigorosa di eventuali duplicati per corsa
trip_full = trip_full.drop_duplicates(subset=['trip_id'])

output_cols = ['linea', 'trip_id', 'trip_headsign', 'fermata_partenza', 'departure_time', 'fermata_arrivo', 'arrival_time', 'giorni_validita', 'start_date', 'end_date']
df_corse = trip_full[[c for c in output_cols if c in trip_full.columns]]

# ==========================================
# 2. GEOLOCALIZZAZIONE BLINDATA CON ITALIA.GEOJSON
# ==========================================
print("🌍 [2/4] Geolocalizzazione fermate tramite 'Italia.geojson'...")

stops_clean = stops_unique[['stop_id', 'stop_name', 'stop_lat', 'stop_lon']].copy()
stops_clean['comune'] = "Non disponibile"
stops_clean['provincia'] = "Non disponibile"

if os.path.exists(GEOJSON_FILE):
    try:
        gdf_stops = gpd.GeoDataFrame(
            stops_clean.copy(), 
            geometry=gpd.points_from_xy(stops_clean.stop_lon, stops_clean.stop_lat),
            crs="EPSG:4326"
        )
        gdf_comuni = gpd.read_file(GEOJSON_FILE)
        if gdf_comuni.crs != "EPSG:4326":
            gdf_comuni = gdf_comuni.to_crs("EPSG:4326")
        
        # Spatial join e rimozione duplicati sui confini
        fermate_geo = gpd.sjoin(gdf_stops, gdf_comuni, how="left", predicate="within")
        fermate_geo = fermate_geo.drop_duplicates(subset=['stop_id'])
        
        col_comune = next((c for c in ['COMUNE', 'comune', 'Nome_Com', 'COM_NAME', 'DEN_COM'] if c in fermate_geo.columns), None)
        col_prov = next((c for c in ['SIGLA', 'DEN_PROV', 'PROVINCIA', 'provincia', 'Sigla Prov', 'COD_PROV', 'SIGLA_PROV', 'PRO_PROV'] if c in fermate_geo.columns), None)
        
        cols_to_merge = ['stop_id']
        rename_dict = {}
        
        if col_comune:
            cols_to_merge.append(col_comune)
            rename_dict[col_comune] = 'comune_val'
        if col_prov:
            cols_to_merge.append(col_prov)
            rename_dict[col_prov] = 'provincia_val'
            
        sub_geo = fermate_geo[cols_to_merge].rename(columns=rename_dict)
        
        # Unione sicura su stop_id per evitare perdite di dati
        stops_clean = pd.merge(stops_clean[['stop_id', 'stop_name', 'stop_lat', 'stop_lon']], sub_geo, on='stop_id', how='left')
        
        stops_clean['comune'] = stops_clean['comune_val'].fillna('Non disponibile') if 'comune_val' in stops_clean.columns else 'Non disponibile'
        stops_clean['provincia'] = stops_clean['provincia_val'].fillna('Non disponibile') if 'provincia_val' in stops_clean.columns else 'Non disponibile'
        
        # Pulizia finale colonne
        stops_clean = stops_clean[['stop_id', 'stop_name', 'stop_lat', 'stop_lon', 'comune', 'provincia']]
        print("   -> Geolocalizzazione completata con successo!")
    except Exception as e:
        print(f"   -> Avviso GeoJSON: {e}")
else:
    print(f"   -> File '{GEOJSON_FILE}' non trovato nella cartella.")

df_fermate_excel = stops_clean[['stop_id', 'stop_name', 'comune', 'provincia']].copy()

# ==========================================
# 3. SALVATAGGIO FILE EXCEL
# ==========================================
print("💾 [3/4] Salvataggio file Excel...")
with pd.ExcelWriter(OUTPUT_EXCEL, engine='openpyxl') as writer:
    df_corse.to_excel(writer, sheet_name='Riepilogo Corse', index=False)
    df_fermate_excel.to_excel(writer, sheet_name='Fermate', index=False)

print(f"✅ File Excel creato con successo: {OUTPUT_EXCEL}")

# ==========================================
# 4. DASHBOARD INTERATTIVA (Base Pulita)
# ==========================================
print("📊 [4/4] Inizializzazione Dashboard Interattiva...")

linee_disponibili = sorted(df_corse['linea'].dropna().unique().tolist())
dropdown_linea = widgets.Dropdown(
    options=linee_disponibili,
    description='Seleziona Linea:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='350px')
)

out_map = widgets.Output()
out_table = widgets.Output()

def aggiorna_dashboard(change):
    linea_scelta = change['new']
    
    with out_map:
        clear_output(wait=True)
        trips_linea = df_corse[df_corse['linea'] == linea_scelta]['trip_id'].tolist()
        
        if trips_linea:
            stop_ids_linea = stop_times[stop_times['trip_id'].isin(trips_linea)]['stop_id'].unique()
            fermate_linea = stops_clean[stops_clean['stop_id'].isin(stop_ids_linea)]
            
            if not fermate_linea.empty:
                centro_lat = fermate_linea['stop_lat'].mean()
                centro_lon = fermate_linea['stop_lon'].mean()
                
                m = folium.Map(location=[centro_lat, centro_lon], zoom_start=12, tiles="Esri.WorldStreetMap")
                
                for t_id in trips_linea[:15]:
                    st_t = stop_times[stop_times['trip_id'] == t_id].sort_values('stop_sequence')
                    p_stops = pd.merge(st_t, stops_clean, on='stop_id')
                    if len(p_stops) > 1:
                        coords = p_stops[['stop_lat', 'stop_lon']].values.tolist()
                        folium.PolyLine(coords, color="blue", weight=4, opacity=0.6).add_to(m)
                
                for _, row in fermate_linea.iterrows():
                    popup_text = f"<b>{row['stop_name']}</b><br>ID: {row['stop_id']}<br>Comune: {row['comune']}<br>Provincia: {row['provincia']}"
                    folium.Marker(
                        location=[row['stop_lat'], row['stop_lon']],
                        popup=popup_text,
                        tooltip=row['stop_name'],
                        icon=folium.Icon(color='red', icon='bus', prefix='fa')
                    ).add_to(m)
                
                m.save("mappa_percorso_linea.html")
                display(m)
            else:
                print("Nessuna fermata trovata per questa linea.")
        else:
            print("Nessuna corsa trovata per la linea selezionata.")

    with out_table:
        clear_output(wait=True)
        corse_linea = df_corse[df_corse['linea'] == linea_scelta]
        display(corse_linea.head(15))

dropdown_linea.observe(aggiorna_dashboard, names='value')

if linee_disponibili:
    aggiorna_dashboard
```
